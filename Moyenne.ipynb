Version1

```python
def calculemoyenneUE():
    moyenne=0
    somme=0
    NBnote=input("combien de notes avez-vous ?")
    for i in range(101,101+int(NBnote)):
        note=input(f"inserez la note de la ressource R{i}")
        coef=input(f"inserez le coef de la ressource")
        moyenne+=float(note)*float(coef)
        somme+=float(coef)
    return "la moyenne de l'UE est: "+str(round((moyenne/somme), 2))
```

```python
calculemoyenneUE()
```

```python
import numpy as np
import matplotlib.pyplot as plt

matiere = [f"R{i}" for i in range(101, 116)] + [f"SAE{i}" for i in range(11, 17)]

note = [float(input(f"Quelle est la note pour {elt} ? ")) for elt in matiere]

coef1 = [10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coef2 = [4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coef3 = [4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

def moy(coef):
    return round(np.average(note, weights=coef), 3)

moyennes = [moy(coef1), moy(coef2), moy(coef3)]

print(f"Les moyennes des UE sont :")
print(f"UE1 = {moyennes[0]}")
print(f"UE2 = {moyennes[1]}")
print(f"UE3 = {moyennes[2]}")

def couleur(val):
    if val >= 10:
        return "green"
    elif val >= 8:
        return "orange"
    else:
        return "red"

colors = [couleur(val) for val in moyennes]
plt.bar(["UE1", "UE2", "UE3"], moyennes, color=colors)
plt.ylim(0, 20)
plt.ylabel("Moyenne")
plt.title("Moyennes des trois UE")
plt.show()
```
