```python
def calculemoyenneUE():
    moyenne=0
    somme=0
    NBnote=input("combien de notes avez-vous ?")
    for i in range(101,101+int(NBnote)):
        note=input(f"inserez la note de la ressource R{i}")
        coef=input(f"inserez le coef de la ressource")
        moyenne+=float(note)*float(coef)
        somme+=float(coef)
    return "la moyenne de l'UE est: "+str(round((moyenne/somme), 2))
```

```python
calculemoyenneUE()
```

```python
def moy_UE1():
   liste_coef=[10,10,7,7,5,6,5,4,2,5,5,20,20,7]
   coef_tot=0
   moy = 0 
   valeur = int(input("Indiquez le nombre de notes que vous avez"))
   liste_note = []
   for k in range(valeur):
       liste_note.append(int(input("Indiquez ici vos notes dans un ordre respectif de vos coef")))
   for i in range(valeur):
       moy += liste_note[i] * liste_coef[i]
       coef_tot += liste_coef[i]
   return moy / coef_tot

moy_UE1()
```

```python
def calculmoyenneDico():
    coef={"R101":10,"R102":10,"R103":7}
    coeftotal=0
    moy=0
    valeur=int(input("NBnote"))
    listenote={}
    for k in range(101,101+valeur):
        ressource=input("quel est le nom de  la ressource ?")
        if ressource not in listenote:
            listenote[ressource]=int(input("note"))
    for i in range(valeur):
        moy+=listenote[f"R{i+101}"]*coef[f"R{i+101}"]
        coeftotal+=coef[f"R{i+101}"]
    return "la moyenne de l'UE est: "+str((round((moy/coeftotal), 2)))
            
calculmoyenneDico()
```
